```python
import plotly.graph_objects as go
import numpy as np
# Create x values (e.g., from 0 to 100 to show the intersection)
x = np.linspace(0, 100, 200)
# Define the two equations
y1 = 1.5 * x
y2 = 10 * (100 - x)
# Initialize the figure
fig = go.Figure()
# Add the first line (y1 = 1.5x)
fig.add_trace(go.Scatter(
    x=x, 
    y=y1, 
    mode='lines', 
    name='y1 = 1.5x'
))
# Add the second line (y2 = 10*(100-x))
fig.add_trace(go.Scatter(
    x=x, 
    y=y2, 
    mode='lines', 
    name='y2 = 10*(100-x)'
))
# Add title and labels (optional, but good for context)
fig.update_layout(
    title='Comparison of Two Functions',
    xaxis_title='x',
    yaxis_title='y'
)
# Display the plot
fig.show()
```
