```python
import matplotlib.pyplot as plt
import pickle
import numpy as np
train_x = pickle.load(open("MNIST_train_x.pkl", "rb"))/255.0
train_y = pickle.load(open("MNIST_train_y.pkl", "rb"))
test_x = pickle.load(open("MNIST_test_x.pkl", "rb"))/255.0
test_y = pickle.load(open("MNIST_test_y.pkl", "rb"))
print("training data: ", train_x.shape, "label: ", train_y.shape)
print("testing data: ", test_x.shape, "label: ", test_y.shape)
```

- Dataset is split into 2 subsets:
- - Training data (usually 60,000)
- - Test data (usually 10,000)
Each image is composed of grayscale pixels: table of 28×28 pixel values, which are “flattened” to vector of 784 values (thus 1 image is 1 row of input data with 784 columns)

```python
def plotExamples(data, label, model_predict):
    plt.figure(figsize=(10, 6))
    for i in range(8):
        sub = 241 + i
        ax = plt.subplot(sub)
        index = np.random.randint(0, data.shape[0])
        ax.set_title("label: " + str(np.argmax(label[index])) + " predict: " + str(np.argmax(model_predict[index])))
        im = np.reshape(data[index], (28, 28))
        plt.imshow(im, cmap="gray")
    plt.show()

plotExamples(train_x, train_y, train_y) # prediction == label == train_y
```
