```python
from mega import Mega

def simple_mega_client(login_email=None, login_password=None):
    # 1. Initialization
    # We create a 'Mega' object, which acts similarly to the 'MegaApi' object in the C++ client.
    mega = Mega()

    # 2. Authentication
    print("Logging into MEGA...")
    email = login_email
    password = login_password

    try:
        # This authenticates your session and fetches your file tree
        m = mega.login(email, password)
        print("Successfully logged in!\n")
    except Exception as e:
        print(f"Login failed: {e}")
        return

    # 3. List Files ('ls' equivalent)
    print("--- Your MEGA Files ---")
    # get_files() retrieves the node tree (files and folders)
    files = m.get_files()
    
    for file_id, file_info in files.items():
        # The 'a' dictionary contains the file attributes, and 'n' is the name
        file_name = file_info['a']['n']
        print(f"- {file_name}")
    print("-----------------------\n")

    # 4. Upload a File ('put' equivalent)
    local_file_path = "gm.pik"
    
            
    print(f"Uploading '{local_file_path}' to MEGA...")
    try:     
        # m.upload() handles encrypting and sending the file to your root directory
        uploaded_file = m.upload(local_file_path)
        print(f"Upload complete! File node details: {uploaded_file}")
    except Exception as e:
        print(f"Upload failed: {e}")



simple_mega_client(login_email="my@yahoo.com", login_password="mypassword")
```

```python
# this is troubleshooting code to debug the login issue. ignore it if login succeeds
import logging
import sys
from mega import Mega

# 1. Enable Debugging
# This tells the 'requests' library (which mega.py uses) to print 
# every detail of the conversation with the MEGA servers.
logging.basicConfig(level=logging.DEBUG)

def troubleshoot_mega_login(email, password):
    mega = Mega()
    
    print(f"--- Starting Troubleshooting for {email} ---")
    
    try:
        # 2. Attempt Login
        # The library tries to POST to https://g.api.mega.co.nz/
        m = mega.login(email, password)
        print("Login Successful!")
        return m
        
    except Exception as e:
        print("\n--- ERROR ANALYSIS ---")
        print(f"Error Type: {type(e).__name__}")
        print(f"Error Message: {e}")
        
        print("\nPossible Causes:")
        print("1. IP Block: MEGA might be blocking your current network (common in Colab/Cloud).")
        print("2. 2FA: If you have Two-Factor Auth enabled, this library will fail with this error.")
        print("3. Maintenance: The MEGA API endpoint might be temporarily down.")
        
        # This helps us see if we got an empty string vs an HTML page
        if "Expecting value" in str(e):
            print("\nTechnical Detail: The server sent a non-JSON response.")
            print("Check the 'DEBUG' logs above for a 403 (Forbidden) or 500 (Error) status code.")



troubleshoot_mega_login("username", "password")
```
